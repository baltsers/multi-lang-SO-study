```python
from fuzzywuzzy import fuzz
from fuzzywuzzy import process
import pandas as pd
from pandas import DataFrame
import re
```

```python
def CleanText(Text):
    Text = str(Text)
    Text = re.sub(r'[+|/]', ' and ', Text)
    Text = re.sub(r'[^\w\d,]', ' ', Text)
    Text = Text.lower()
    words = Text.split()
    words = [re.sub(r'[^a-z]', '', word) for word in words if word.isalnum()]
    min_len = 4
    words = [word for word in words if len(word) > min_len]
    return words
```

```python
# sample = pd.read_csv('sample50.csv')
sample = pd.read_csv('data_all_QA.csv')
sample.head()
```

```python
topics = pd.read_csv('strongTopics_words.csv')
topics.head()
```

```python
pred = []
for index, rowSample in sample.iterrows():
    multi = 'N'
    k = 0
    for index, rowTopic in topics.iterrows():
        if multi == 'N' and k<2:
            if rowTopic['type'] == 'single':
                score = process.extractOne(rowTopic['topic1'], CleanText(rowSample['question
'])+CleanText(rowSample['solution']))#scorer=fuzz.token_sort_ratio
                if score[1]>90:
#                     multi='Y'
                    k+=1
                    print(rowTopic['type'])
                    print(score)
            elif rowTopic['type'] == 'double':
                score1 = process.extractOne(rowTopic['topic1'], CleanText(rowSample['issues being discuss'])+CleanText(rowSample['solution']))
                score2 = process.extractOne(rowTopic['topic2'], CleanText(rowSample['issues being discuss'])+CleanText(rowSample['solution']))
                if score1[1]>90 and score2[1]>90:
                    multi='Y'
                    print(rowTopic['type'])
                    print(score1,score2)
        else:
            print(k)
            multi='Y'
            break
    pred.append(multi)
```

```python
result = sample
result['predict_multi']=pred
# result.to_csv('pred_sample.csv')
result.to_csv('new_data_2008-2018_filtered.csv')
```

```python
TP,TN,FP, FN = 0,0,0,0
for index, row in result.iterrows():
    if(row['predict_multi'] == row['Multi']):
        if(row['predict_multi'] == 'Y'):
            TP+=1
        else:
            TN+=1
    else:
        if(row['predict_multi'] == 'Y'):
            FP+=1
        else:
            FN+=1
            
print(TP,TN,FP,FN)
```

```python
accuracy = (TP+TN)/(TP+TN+FP+FN)
precision = TP/(TP+FP)
recall = TP/(TP+FN)
print('accuracy:',accuracy)
print('precision:',precision)
print('recall:',recall)
```
